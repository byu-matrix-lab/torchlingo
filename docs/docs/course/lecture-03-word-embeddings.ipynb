## Step 2. Load the model

First run downloads about 517 MB and takes roughly 20 to 40 seconds on Colab.
It is cached after that, so re-running this cell is instant.

`distiluse-base-multilingual-cased-v2` maps text from 50+ languages into one
512-dimensional space. That shared space is the whole point of today's lecture.

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("distiluse-base-multilingual-cased-v2")
print("loaded. embedding dimension:", model.get_sentence_embedding_dimension())
```

## Step 3. Your words

**Edit this cell.** Replace the Spanish with your own language, keeping the order
aligned: `mine[i]` must be the translation of `eng[i]`.

Pick everyday words. Concrete nouns and common verbs behave best.
If your language is not in the 50, pair up with someone whose language is,
then note which language you wanted in the discussion at the end.

```python
eng  = ["water", "fire",  "mother", "to run", "book"]
mine = ["agua",  "fuego", "madre",  "correr", "libro"]

language_name = "Spanish"   # change me

assert len(eng) == len(mine), "the two lists must be the same length"
print(f"{len(eng)} pairs, English to {language_name}")
```

## Step 4. Embed and compare

`util.cos_sim(A, B)` returns a matrix where entry `[i][j]` is the cosine similarity
between `A[i]` and `B[j]`. Row `i` is one English word compared against every word
in your language, so the translation you intended sits on the diagonal.

**Read the matrix yourself.** Is the diagonal the largest value in its row, or does
something else win?

```python
E = model.encode(eng + mine)
S = util.cos_sim(E[:len(eng)], E[len(eng):]).numpy()

w = max(len(x) for x in mine) + 2
print(" " * 10 + "".join(f"{m:>{w}}" for m in mine))
for i, e in enumerate(eng):
    print(f"{e:>9} " + "".join(f"{S[i][j]:>{w}.2f}" for j in range(len(mine))))
```
